```python
%%sql
Create table Super_Bowl (
Product_id INTEGER PRIMARY KEY,
product_name text,
product_price REAL
) STRICT;
```

```python
%%sql
insert into Super_bowl (product_id, product_name, product_price)
values
(1,'Super Bowl Jersey', 120),
(2, 'Super Bowl Hat', 45),
(3, 'Super Bowl Tshirt', 75);
```

```python
%%sql
update Super_Bowl set product_price = 100 where product_name = 'Super Bowl Jersey'
```
